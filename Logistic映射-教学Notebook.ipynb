# Logistic Map · Sustainable Fishing Model
One equation shows how a fish population moves from **stable** all the way to **chaos** — and answers: can we fish it sustainably?

Maps to 3 lessons: ① fish game (unplugged) → ② code it → ③ tune r to see bifurcation & chaos.

## Step 1 · Start with the naive idea: exponential growth
If fish only reproduce, with no limits, what happens? Try the 'reproduce-only' model: P(t+1) = r·P(t).

```python
import matplotlib.pyplot as plt

r = 1.5
P = [0.1]
for t in range(30):
    P.append(r * P[t])

plt.plot(P, 'o-')
plt.title('Exponential: P(t+1) = r*P(t)  —  grows forever (unrealistic!)')
plt.xlabel('Generation'); plt.ylabel('P'); plt.show()
```

Run it: the fish grow forever — impossible in reality, because a pond has a **carrying capacity**.
**CT · evaluation**: the model gives an unrealistic result, so we go back and fix it.

## Step 2 · Add carrying capacity: (1 − P)
P is the fraction of carrying capacity. As P approaches 1 (pond full), growth slows. This is the **Logistic Map**:
P(t+1) = r · P(t) · (1 − P(t))
**CT · abstraction**: a whole pond of fish → one number P.

```python
r = 2.8
P = [0.1]
for t in range(40):
    P.append(r * P[t] * (1 - P[t]))

plt.plot(P, '.-')
plt.axhline(1 - 1/r, color='red', ls='--', label=f'P* = {1-1/r:.3f}')
plt.legend()
plt.title('Logistic Map: r = 2.8  →  stable P*')
plt.xlabel('Generation'); plt.ylabel('P'); plt.show()
```

## Step 3 · Turn r up and see what happens
**CT · modelling / algorithms**: change one parameter and watch how the system behaves.

```python
fig, axes = plt.subplots(1, 4, figsize=(15, 3.5))
for ax, r in zip(axes, [2.8, 3.2, 3.5, 3.9]):
    P = [0.1]
    for t in range(50):
        P.append(r * P[t] * (1 - P[t]))
    ax.plot(P, '.-', lw=1)
    ax.set_title(f'r = {r}'); ax.set_ylim(0, 1)
plt.tight_layout(); plt.show()
```

## Step 4 · One picture, the whole story: the bifurcation diagram
x-axis = r, y-axis = 'where it settles long-term'. Stable → splits into 2 → 4 → a cloud = **chaos**.

```python
rs, Ps = [], []
for r in [i/1000 for i in range(2500, 4001)]:
    p = 0.5
    for _ in range(300): p = r*p*(1-p)
    for _ in range(100):
        p = r*p*(1-p); rs.append(r); Ps.append(p)

plt.figure(figsize=(10, 6))
plt.scatter(rs, Ps, s=0.15, c='#2563eb')
plt.xlabel('r'); plt.ylabel('P (long-term)')
plt.title('Bifurcation diagram: stable → period-doubling → chaos'); plt.show()
```

## Step 5 · The butterfly effect: a tiny difference at the start, a totally different ending
**CT · evaluation**: a deterministic equation ≠ predictable.

```python
r = 3.9
for P0 in [0.1, 0.10001]:
    P = [P0]
    for t in range(40):
        P.append(r * P[t] * (1 - P[t]))
    plt.plot(P, label=f'P0 = {P0}')

plt.legend()
plt.title('Butterfly effect: P0 differs by 0.00001')
plt.xlabel('Generation'); plt.ylabel('P'); plt.show()
```

## Step 6 · A real application: sustainable fishing
A fishery harvests a fixed amount h each year. How much is too much?
P(t+1) = r · P(t) · (1 − P(t)) − h

```python
r = 2.8
for h in [0, 0.1, 0.3]:
    P = [0.3]
    for t in range(40):
        P.append(max(0, r * P[t] * (1 - P[t]) - h))
    plt.plot(P, label=f'h = {h}')

plt.legend()
plt.title('Harvesting: too much h crashes the population')
plt.xlabel('Generation'); plt.ylabel('P'); plt.show()
```

## Step 7 · What does this model leave out? (critical reflection)
- **Closed pond**: no migration, no predators, no seasons
- **Deterministic**: no random events (weather, disease)
- **Synchronised generations**: real populations are continuous, with age structure

**CT · evaluation**: all models are wrong, but some are useful. The key is knowing what they leave out.
